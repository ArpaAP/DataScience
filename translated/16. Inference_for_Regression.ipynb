# 회귀에 대한 추론

지금까지 우리의 변수 간 관계 분석은 순전히 기술적이었습니다. 우리는 산점도를 통해 그릴 수 있는 최적의 직선을 찾는 방법을 알고 있습니다. 이 선은 모든 직선 중에서 추정의 평균 제곱 오차가 가장 작다는 의미에서 최적입니다.

그러나 우리의 데이터가 더 큰 모집단의 표본일 뿐이라면 어떻게 될까요? 표본에서 두 변수 간에 선형 관계를 발견했다면, 모집단에서도 같을까요? 정확히 동일한 선형 관계일까요? 표본에 없는 새로운 개인의 반응을 예측할 수 있을까요?

이러한 추론과 예측의 질문은 산점도가 플로팅되는 두 변수 간의 기저 관계를 반영하지만 그 관계를 완전히 명시하지 않는다고 믿을 때 발생합니다. 예를 들어, 재태 기간 대비 출생 체중의 산점도는 표본에서 두 변수 간의 정확한 관계를 보여줍니다. 하지만 그 관계가 표본이 추출된 모집단의 모든 아기들, 또는 실제로 일반적인 아기들에게 참인지, 또는 거의 참인지 궁금할 수 있습니다.

항상 그렇듯이, 추론적 사고는 데이터에 대한 가정을 신중하게 검토하는 것으로 시작합니다. 가정의 집합을 *모델(models)*이라고 합니다. 대략 선형인 산점도에서 무작위성에 대한 가정의 집합을 *회귀 모델(regression models)*이라고 합니다.
