# 평균이 중요한 이유
이 과정에서 우리는 총변동거리, 최댓값, 중앙값, 그리고 평균을 포함한 여러 가지 통계량을 공부해왔습니다. 무작위성에 대한 명확한 가정 하에서, 우리는 이러한 모든 통계량의 경험적 분포를 그려왔습니다. 최댓값이나 총변동거리 같은 일부 통계량은 한쪽 방향이나 다른 방향으로 명확하게 치우친 분포를 가집니다. 하지만 표본평균의 경험적 분포는 연구하는 모집단에 관계없이 거의 항상 종 모양에 가깝게 나타났습니다.

무작위 표본의 어떤 성질이 *모집단에 관계없이* 참이라면, 그것은 추론을 위한 강력한 도구가 됩니다. 왜냐하면 우리는 전체 모집단의 데이터에 대해 많이 알지 못하기 때문입니다. 큰 무작위 표본의 평균 분포가 바로 이러한 성질의 범주에 속합니다. 그래서 무작위 표본평균이 데이터 과학에서 광범위하게 사용됩니다.

이 장에서는 기초 모집단에 대한 최소한의 가정만으로 평균과 평균에 대해 말할 수 있는 것들을 공부할 것입니다. 다룰 질문들은 다음과 같습니다:

- 평균은 정확히 무엇을 측정하는가?
- 대부분의 데이터는 평균에 얼마나 가까운가?
- 표본 크기는 표본평균의 변동성과 어떤 관계가 있는가?
- 왜 무작위 표본평균의 경험적 분포는 종 모양으로 나타나는가?
- 추론을 위해 표본평균을 어떻게 효과적으로 사용할 수 있는가?
